### 2.2 Importing Libraries and Loading the Dataset
With the file verified, the necessary PySpark library was imported and reads the CSV into a Spark DataFrame:

### 2.2.1 Importing PySpark functions and reading the CSV

The import statement brings in PySpark's built-in column functions under the alias F, making functions such as split(), explode(), and months_between() available throughout the notebook. The spark.read.csv() call reads the file into a distributed Spark DataFrame using the following options:
- header=True tells Spark that the first row contains column names rather than data.
- inferSchema=True instructs Spark to scan the data and automatically assign the most appropriate data type to each column, such as DateType for the date columns and IntegerType for the Enrollment column.
- quote and escape are both set to the double-quote character because some fields in the file are wrapped in quotes and may contain commas or newlines inside them. These options tell the parser where a field begins and ends.
- multiLine=True allows Spark to correctly parse fields that span more than one line, which occurs in some study titles in this dataset.

After loading, four inspection commands were run. display(10) renders an interactive table of the first 10 rows in the Databricks UI. printSchema() prints the column names and their inferred data types. show(5) prints a plain-text table to the console output. count() returns the total number of rows, which confirmed that 570,798 records were loaded successfully.


```python
# Import PySpark's built-in functions

from pyspark.sql import functions as F

# Define the file path to the dataset in Unity Catalog
filePath = '/Volumes/teaching/datasets/assignment_2/task_1/ctg-studies.csv'

ClinicalDf = spark.read.csv( 
    filePath,
    header=True,   # First row contains column names
    inferSchema=True,  # Automatically detect data types (dates, integers, etc.)
    quote='"', # Fields wrapped in double quotes are treated as a single value
    escape='"', # A double quote inside a quoted field is escaped with another quote
    multiLine=True # Allow fields that span multiple lines
)

ClinicalDf.display(10, truncate=80) # Display the first 10 rows as an interactive table in Databricks
ClinicalDf.printSchema()
ClinicalDf.show(5) 
ClinicalDf.count() # Count the total number of rows to confirm the full dataset is loaded
```

### 2.2.2 Renaming the columns

The original column headers contain spaces which cause errors when used directly in SQL. Each column is renamed using withColumnRenamed(), replacing spaces with underscores. Because Spark DataFrames are immutable, each call returns a new DataFrame and the result is reassigned back to ClinicalDf. printSchema() and display() are called again afterwards to confirm the renaming was applied correctly.

```python
# Rename all columns to remove spaces, replacing them with underscores

ClinicalDf = ClinicalDf.withColumnRenamed("NCT Number","nct_number")
ClinicalDf = ClinicalDf.withColumnRenamed("Study Title","study_title")
ClinicalDf = ClinicalDf.withColumnRenamed("Acronym", "acronym")
ClinicalDf = ClinicalDf.withColumnRenamed("Study Status","study_status")
ClinicalDf = ClinicalDf.withColumnRenamed("Conditions","conditions")
ClinicalDf = ClinicalDf.withColumnRenamed("Interventions","interventions")
ClinicalDf = ClinicalDf.withColumnRenamed("Sponsor","sponsor")
ClinicalDf = ClinicalDf.withColumnRenamed("Collaborators","collaborators")
ClinicalDf = ClinicalDf.withColumnRenamed("Enrollment","enrollment")
ClinicalDf = ClinicalDf.withColumnRenamed("Funder Type","funder_type")
ClinicalDf = ClinicalDf.withColumnRenamed("Study Type","study_type")
ClinicalDf = ClinicalDf.withColumnRenamed("Study Design","study_design")
ClinicalDf = ClinicalDf.withColumnRenamed("Start Date","start_date")
ClinicalDf = ClinicalDf.withColumnRenamed("Completion Date", "completion_date")

# Confirm the new column names and data types
ClinicalDf.printSchema()
display(ClinicalDf.limit(10))
```

### 2.2.3 Checking for unique records

This step verifies that the nct_number column, the unique identifier for each clinical trial, contains no duplicates. It does this by comparing the count of distinct nct_number values against the total row count. If both numbers are equal, the result is True, confirming that every row represents a different trial. This matters because it means that COUNT(*) in later queries accurately reflects the number of unique trials rather than counting duplicates.

```python
# Check whether every row has a unique nct_number

is_unique = (
    ClinicalDf.select("nct_number")
      .distinct()
      .count()
    == ClinicalDf.count()  # Compare distinct count to total row count
)
print(is_unique)
ClinicalDf.count()
```

## 3. Analysis

With the data loaded and cleaned, the four questions are answered with Spark SQL. A row count is run first to confirm the total number of records available.

```python
%sql
-- Count the total number of rows to confirm the full dataset is available

SELECT COUNT(*) AS total_rows
FROM ClinicalTrials
```

### 3.1 Clinical Trial Types by Frequency

Which types of clinical trial are most common? GROUP BY on study_type with COUNT(*) gives the frequency of each type. COALESCE labels any NULL values as Unknown, and ORDER BY frequency DESC sorts the results from most to least frequent.

```python
%sql
SELECT
    COALESCE(study_type, 'Unknown') AS study_type, -- Replace NULL values with 'Unknown'
    COUNT(*)                        AS frequency -- Count trials per study type
FROM ClinicalTrials
GROUP BY study_type -- Group rows by each distinct study type
ORDER BY frequency DESC -- Sort from most to least frequent
```

### 3.2 Top 10 Conditions by Frequency

The conditions column can contain several conditions in one row, separated by a pipe (|) character, for example Diabetes|Obesity|Hypertension. To count each condition individually, a three-step process is used with temporary views.

Step 1 : Split the conditions column into an array:

The split() function breaks the conditions string apart at every pipe character and returns the pieces as an array. The delimiter is written as \| because the pipe is a special character in regular expressions and must be escaped. The result is stored as a temporary view called split_conditions. A preview query was run to confirm the output:

```python
%sql

-- Create a temporary view that splits the pipe-separated conditions into an array
CREATE OR REPLACE TEMP VIEW split_conditions AS
SELECT
    nct_number,
    split(conditions, '\\|') AS conditions_array -- Split on | (escaped as \| in regex)
FROM ClinicalTrials
```

```python
%sql
-- Preview the split results to confirm conditions are now stored as arrays

SELECT *
FROM split_conditions
LIMIT 5
```

Step 2 : Explode the array into individual rows

The explode() function takes each array produced in the previous step and turns every element into its own separate row. For example, a trial with three conditions in its array will produce three rows in this view, one per condition. This is what allows each condition to be counted individually. A preview was run again to verify the output:

```python
%sql

-- Create a temporary view that turns each array element into its own row

CREATE OR REPLACE TEMP VIEW exploded_conditions AS
SELECT
    nct_number,
    explode(conditions_array) AS condition -- Each condition becomes a separate row
FROM split_conditions
```

```python
%sql

-- Preview to confirm each condition now occupies its own row
SELECT *
FROM exploded_conditions
LIMIT 5
```

Step 3 : Count and retrieve the top 10

TRIM() removes any extra whitespace from the beginning or end of each condition name. The WHERE clause filters out NULL values and empty strings that can appear when a conditions string ends with a pipe or contains consecutive pipes. GROUP BY condition and COUNT(*) count how many times each condition appears across all trials. ORDER BY frequency DESC and LIMIT 10 return only the ten most common conditions.

```python
%sql
SELECT
    TRIM(condition) AS condition, -- Remove any leading or trailing whitespace
    COUNT(*) AS frequency  -- Count how many times each condition appears
FROM exploded_conditions  
WHERE condition IS NOT NULL  -- Exclude rows where condition is NULL
AND condition != '' -- Exclude empty strings from trailing pipe characters
GROUP BY condition  -- Group by each unique condition name
ORDER BY frequency DESC  -- Sort from most to least frequent
LIMIT 10  -- Return only the top 10 results
```

### 3.3 Mean Clinical Trial Length in Months

For studies with an end date, what is the mean trial length in months?

months_between(completion_date, start_date) calculates the number of months between the two dates as a decimal. This is more accurate than dividing the number of days by 30, because it accounts for months of different lengths. AVG() computes the mean across all qualifying rows, and ROUND(..., 2) formats the result to two decimal places. The WHERE clause keeps only rows where both dates are present and the completion date falls after the start date, removing records that appear to contain data entry errors.

```python
%sql
SELECT
    ROUND(AVG(months_between(completion_date, start_date)), 2) AS mean_duration_months,
    -- months_between() calculates the exact months between two dates
    -- More accurate than dividing days by 30 as it accounts for varying month lengths
    COUNT(*) AS number_of_trials
FROM ClinicalTrials
WHERE start_date IS NOT NULL
AND   completion_date IS NOT NULL
AND   completion_date > start_date  -- Remove records where end date is before start (data errors)
```

### 3.4 Alzheimer's Disease Trial Trend Over Time

How many completed Alzheimer's disease trials finished each year? Only studies with a non-null completion date and a status of Completed are counted.

YEAR(completion_date) extracts the year so trials can be grouped by the year they finished. The WHERE clause applies three filters: completion_date IS NOT NULL keeps trials with a recorded completion date; study_status = 'COMPLETED' keeps only fully completed trials; and the LIKE conditions match any trial whose conditions mention Alzheimer or alzheimer. GROUP BY and ORDER BY produce a chronological count, which was displayed as a line chart in Databricks with the year on the x-axis and the number of trials on the y-axis.

```python
%sql
SELECT
    YEAR(completion_date) AS completion_year, -- Extract the year from the completion date
    COUNT(*) AS number_of_trials   -- Count trials completed in each year
FROM ClinicalTrials
WHERE completion_date IS NOT NULL   -- Only include trials with a completion date
AND study_status  =  'COMPLETED'   -- Only include fully completed trials
AND (conditions LIKE '%Alzheimer%' OR conditions LIKE '%alzheimer%') -- Match trials mentioning Alzheimer (capital A) or alzheimer (lowercase a)
GROUP BY completion_year  -- Group the count by year
ORDER BY completion_year ASC
```
